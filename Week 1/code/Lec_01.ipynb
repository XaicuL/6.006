# **점근 표기법** - 시간 복잡도 클래스별 실제 측정

## 실험 1: $Θ(1), Θ(log n), Θ(n), Θ(n log n), Θ(n²)$의 실제 증가 패턴

**검증 목표**
강의 노트의 시간 복잡도 표(n=1000일 때 1ns~1ms)에 나온 <br>
각 복잡도 클래스가 실제로 다른 증가율을 보이는지 확인한다.

**관련 개념 (Why)**
점근 표기법은 "n이 커질수록 지배적인 항"을 기준으로 분류한다. <br>
상수/로그/선형/로그선형/이차 순으로 증가율이 급격히 벌어져야 한다.

**예상 결과 (가설)**
n을 10배씩 늘렸을 때 <br>
- $O(1)$: 시간 거의 그대로
- $O(log n)$: 시간 아주 조금 증가
- $O(n)$: 시간 10배 증가
- $O(n log n)$: 시간 10배보다 조금 더 증가
- $O(n²)$: 시간 100배 증가

**변인 설계**
- 각 복잡도를 흉내내는 더미 연산 함수를 만들어 실행 시간 측정
- n ∈ {1000, 10000, 100000}

```python
import time

def op_constant(n):
    x = 0
    for _ in range(n):
        x += 1

def op_log(n):
    x = 0
    k = n

    while k > 1:
        k // 2
        x += 1

def op_linear(n):
    x = 0
    for _ in range(n):
        x += 1

def op_nlogn(n):
    x = 0
    for _ in range(n):
        k = n
        while k > 1:
            k //= 2
            x += 1

def op_quadratic(n):
    x = 0
    for _ in range(n):
        for _ in range(n):
            x += 1

funcs = {
    "O(1)": op_constant,
    "O(log n)": op_log,
    "O(n)": op_linear,
    "O(n log n)": op_nlogn,
}

for name, f in funcs.items():
    print(f"---{name}---")

    for n in [1000, 10000, 100000]:
        start = time.time()
        f(n)
        print(f"  n={n:7d} | time={time.perf_counter()-start:.6f}s")

print("---O(n²)---")

for n in [100, 1000, 2000]:
    start = time.time()
    op_quadratic(n)
    print(f"  n={n:7d} | time={time.perf_counter()-start:.6f}s")

```
