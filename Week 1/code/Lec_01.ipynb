```python
n = 10

for i in range(n):
    print("A")
    print("B")

# n = 5 => A A A A A B B B B B : 10번 출력
# n = 10 => A A A A A B B B B B : 20번 출력
# n = 100 => A A A A A B B B B B : 200번 출력
```

```python
n = 3

for i in range(n):
    for j in range(n):
        print(i, j)

# When i = 0
# j = 0 => 0 0
# j = 1 => 0 1
# j = 2 => 0 2

# When i = 1
# j = 0 => 1 0
# j = 1 => 1 1
# j = 2 => 1 2

# When i = 2
# j = 0 => 2 0
# j = 1 => 2 1
# j = 2 => 2 2

# 3 x 3 = 9 times

print('-----------------------')

n = 4

for p in range(n):
    for q in range(n):
        print(p, q)

# When p = 0
# q = 0 => 0 0
# q = 1 => 0 1
# q = 2 => 0 2
# q = 3 => 0 3

# When p = 1
# q = 0 => 1 0
# q = 1 => 1 1
# q = 2 => 1 2
# q = 3 => 1 3


# n = 3일때 9번, n = 4일때 16번 출력 즉 서로 제곱 관계임을 알 수 있다.
# => O(n^2)
```

```python
n = 2

for i in range(n):
    for j in range(n):
        for k in range(n):
            print(i, j, k)

#When i = 0
# j = 0 => 0 0 0
# j = 1 => 0 1 0
# j = 2 => 0 2 0
# j = 0 => 0 0 1
# j = 1 => 0 1 1
# j = 2 => 0 2 1

#When i = 1
# j = 0 => 1 0 0
# j = 1 => 1 1 0
# j = 2 => 1 2 0
# j = 0 => 1 0 1
# j = 1 => 1 1 1
# j = 2 => 1 2 1

# 2 x 2 x 2 = 8 times

# => O(n^3)
```

```python
n = 3
count = 0

for i in range(n):
    for j in range(i + 1):
        count += 1
        print(f'count = {count} : {i}, {j}')

# When i = 0
#Then range of j = 1 => count = 1 : 0, 0

#When i = 1
#Then range of j = 2 => count = 2 : 1, 0
#Then range of j = 2 => count = 3 : 1, 1

#When i = 2
#Then range of j = 3 => count = 4 : 2, 0
#Then range of j = 3 => count = 5 : 2, 1
```

```python
n = 16

while n > 1:
    print(n)
    n = n // 2


# When n = 16
    #16, 8, 4, 2


# When n = 32
    #32, 16, 8, 4, 2

#n 은 두배가 되었으나, 출력은 n 대비 +1 되었다.

#=> log2(n)
```

```python
x = 1
n = 32

while x < n:
    print(x)
    x = x * 2

# x의 변화: 1 → 2 → 4 → 8 → 16 → 32
# 출력:      1   2   4   8   16
# x가 32가 되면 32 < 32가 False이므로 출력하지 않고 종료
# 반복 횟수: 5번 = log₂(32)
```

```python
n = 8 # 8번 반복

for i in range(n):
    print("A", i)

x = 1 #log2(8) = 3 번 반복

while x < n:
    print("B", x)
    x = x * 2

# x의 변화: 1 → 2 → 4 → 8
# 출력:      1   2   4   8
# x가 8이 되면 8 < 8이 False이므로 출력하지 않고 종료
# 반복 횟수: 4번 = log₂(8)

# range(8) = 0~7이지만 총 8개 → for문은 8번 실행
# while: 1 → 2 → 4 → 8, 단 8 < 8은 False → 3번 실행
# 총 실행 횟수 = 8 + 3 = 11
```

```python
n = 16

for i in range(n):          # 16번 = n번
    x = 1

    while x < n:            # 각 i마다 4번 = log₂(n)번
        print(i, x)
        x = x * 2

# for 안에 while이 중첩되어 있음
# → 각 i마다 while이 처음부터 다시 실행됨
# → n × log₂(n)
# → O(n log n)
#
# n = 16일 때:
# 16 × 4 = 64번 print

# 순차 실행이면: n + log n
# 중첩 실행이면: n × log n
```

```python
students = [
    ("Alice", "Jan 10"),
    ("Bob", "May 3"),
    ("Charlie", "Jan 10"),
]

# 지금까지 확인한 학생들
record = [
    ("Alice", "Jan 10"),
    ("Bob", "May 3"),
]

# 새로 확인할 학생
name, birthday = students[2]  # Charlie


# record에 있는 학생들을 한 명씩 꺼내서 비교
for old_name, old_birthday in record:

    print(name, "vs", old_name)

    # 생일이 같은가?
    if birthday == old_birthday:
        print("같은 생일 발견!")
        print(name, old_name)

        # 이미 찾았으니 더 비교할 필요 없음
        break
```

```python
students = [
    ("Alice", "Jan 10"),
    ("Bob", "May 3"),
    ("Charlie", "Jan 10"),
]

record = []

for name, birthday in students:

    print("현재 학생:", name)

    for old_name, old_birthday in record:

        print("  비교:", name, "vs", old_name)

        if birthday == old_birthday:
            print("  같은 생일 발견!")
            print(" ", name, old_name)
            break

    record.append((name, birthday))

    print("현재 record:", record)
    print()
```

```python
def birthday_match(students):

    record = []

    for name, birthday in students:

        for old_name, old_birthday in record:

            if birthday == old_birthday:
                return (name, old_name)

        record.append((name, birthday))


    return None


students = [
    ("Alice", "Jan 10"),
    ("Bob", "May 3"),
    ("Charlie", "Jan 10")
]


result = birthday_match(students)

print(result)
```
